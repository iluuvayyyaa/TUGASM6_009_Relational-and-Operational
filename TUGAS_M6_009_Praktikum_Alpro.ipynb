```python
#1 Lowongan Asisten
usia=int(input("Usia: "))
departemen=input("Nama Departemen: ")
diterima=usia>=19 and "Statistika" in departemen
print(f"Kriteria Syarat Pendaftar: {diterima}")
```

```python
#2 Jalur VIP
data_pelanggan=("Budi Alexander Christi", "Premium")
berat_kardus=[50,20.2,30,5]
nama, status_member=data_pelanggan
total_berat=sum(berat_kardus)
print(total_berat)
print(f'Pengiriman Jalur Prioritas VIP: {status_member=="Premium" or total_berat>100}')

data_pelanggan=("Santoso Hermess", "Non Premium")
berat_kardus=[40,20.3,20.5]
nama, status_member=data_pelanggan
total_berat=sum(berat_kardus)
print(total_berat)
print(f'Pengiriman Jalur Prioritas VIP: {status_member=="Premium" or total_berat>100}')
```
